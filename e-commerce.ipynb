```python
 # E-commerce order proceesing:
def process_order(price, quantity, customer_type="regular"):
    purchase = price * quantity

    if customer_type == "regular":
        if purchase >= 10000:
            discount = purchase * 0.10
        else:
            discount = 0

    elif customer_type == "premium":
        if purchase >= 10000:
            discount = purchase * 0.20
        else:
            discount = 0

    final_amount = purchase - discount

    if final_amount >= 5000:
        shipping = 0
    else:
        shipping = 250

    total = final_amount + shipping

    print("Purchase:", purchase)
    print("Discount:", discount)
    print("Final Amount:", final_amount)
    print("Shipping:", shipping)
    print("Total Bill:", total)


price = float(input("Enter product price: "))
quantity = int(input("Enter quantity: "))
customer_type = input("Enter customer type (regular/premium): ")

process_order(price, quantity, customer_type)
```
