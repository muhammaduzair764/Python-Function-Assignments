```python
# Hospital Appointment Billing:

def calculate_bill(doctor_type, lab_test, medicine):

    if doctor_type == "General":
        doctor_fee = 1500
    elif doctor_type == "Specialist":
        doctor_fee = 3000
    elif doctor_type == "Emergency":
        doctor_fee = 5000
    else:
        doctor_fee = 0
    if lab_test:
        lab_fee = 1000
    else:
        lab_fee = 0
    if medicine:
        medicine_fee = 2000
    else:
        medicine_fee = 0
    total = doctor_fee + lab_fee + medicine_fee
    print("Doctor:", doctor_type)
    print("Doctor Fee:", doctor_fee)
    print("Lab Test:", lab_fee)
    print("Medicine:", medicine_fee)
    print("Total:", total)
doctor_type = input("Enter Doctor Type (General/Specialist/Emergency): ")
lab_input = input("Lab Test? (Yes/No): ")
medicine_input = input("Medicine? (Yes/No): ")

lab_test = lab_input == "yes"
medicine = medicine_input == "yes"

calculate_bill(doctor_type, lab_test, medicine)
```
