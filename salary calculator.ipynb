```python
# Employee Salary Calculator

def calculate_salary(name, basic_salary, working_hours):

    normal_hours = 160
    overtime_rate = 500
    overtime_bonus = 5000

    if working_hours > normal_hours:
        overtime_hours = working_hours - normal_hours
    else:
        overtime_hours = 0

    overtime_pay = overtime_hours * overtime_rate

    if overtime_hours > 20:
        bonus = overtime_bonus
    else:
        bonus = 0

    total_salary = basic_salary + overtime_pay + bonus

    print("Employee:", name)
    print("Basic Salary:", basic_salary)
    print("Working Hours:", working_hours)
    print("Overtime Hours:", overtime_hours)
    print("Overtime Pay:", overtime_pay)
    print("Overtime Bonus:", bonus)
    print("Total Salary:", total_salary)


# Input values
name = input("Enter Employee Name: ")
basic_salary = float(input("Enter Basic Salary: "))
working_hours = float(input("Enter Working Hours: "))

calculate_salary(name, basic_salary, working_hours)
```
