```python
# Student grade calculator:
def calculate_result(name, english, math, computer):
    total = english + math + computer
    percentage = (total / 300) * 100

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "Fail"

    print("Student:", name)
    print("Total:", total)
    print("Percentage:",percentage)
    print("Grade:", grade)


name = input("Enter student name: ")
english = int(input("Enter English marks: "))
math = int(input("Enter Math marks: "))
computer = int(input("Enter Computer marks: "))

calculate_result(name, english, math, computer)
```
