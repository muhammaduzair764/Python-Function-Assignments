```python
# Employee bonus calculator:
def calculate_bonus(name, salary, projects, performance):

    # Performance level
    if performance >= 90:
        level = "Excellent"
    elif performance >= 80:
        level = "Very Good"
    elif performance >= 70:
        level = "Good"
    else:
        level = "Needs Improvement"

    # Bonus
    if level == "Excellent" and projects >= 3:
        bonus = salary * 0.20
    elif level == "Very Good" and projects >= 3:
        bonus = salary * 0.15
    elif level == "Good" and projects >=2:
        bonus = salary * 0.10
    else:
        bonus = 0

    final_salary = salary + bonus

    print("Employee:", name)
    print("Performance:", level)
    print("Projects:", projects)
    print("Bonus:", bonus)
    print("Final Salary:", final_salary)


# User Input
name = input("Enter employee name: ")
salary = float(input("Enter salary: "))
projects = int(input("Enter projects completed: "))
performance = int(input("Enter performance score: "))

calculate_bonus(name, salary, projects, performance)
```
