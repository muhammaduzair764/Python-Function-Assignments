```python
# FINAL PROJECT: MINI BANKING SYSTEM:

# Function 1: Check Balance:
def check_balance(balance):
    print("Current Balance:", balance)
# Function 2: Deposit Money:
def deposit(balance, amount):
    if amount > 0:
        balance = balance + amount
        print("Deposit successful!")
        print("New Balance:", balance)
    else:
        print("Invalid amount! Deposit must be greater than 0.")
    return balance
# Function 3: Withdraw Money:
def withdraw(balance, amount):
    if amount > 0 and amount <= balance and (balance - amount) >= 1000:
        balance = balance - amount
        print("Withdrawal successful!")
        print("Remaining Balance:", balance)
    else:
        print("Withdrawal rejected!")
        print("Minimum remaining balance must be 1000.")
    return balance
# Function 4: Loan Eligibility
def calculate_loan_eligibility(age, salary, credit_score):
    if age >= 21 and salary >= 50000 and credit_score >= 650:
        print("Customer is Eligible for Loan.")
    else:
        print("Customer is Not Eligible for Loan.")

# MAIN PROGRAM:
balance = 10000
while True:
    print("\n===== BANK SYSTEM =====")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Loan Eligibility")
    print("5. Exit")

    choice = input("Enter your choice: ")

    # Option 1
    if choice == "1":
        check_balance(balance)

    # Option 2
    elif choice == "2":
        amount = float(input("Enter deposit amount: "))
        balance = deposit(balance, amount)

    # Option 3
    elif choice == "3":
        amount = float(input("Enter withdrawal amount: "))
        balance = withdraw(balance, amount)

    # Option 4
    elif choice == "4":
        age = int(input("Enter your age: "))
        salary = float(input("Enter your salary: "))
        credit_score = int(input("Enter your credit score: "))

        calculate_loan_eligibility(age, salary, credit_score)

    # Option 5
    elif choice == "5":
        print("Thank you for using BANK SYSTEM!")
        print("Program Ended.")
        break

    # Invalid choice
    else:
        print("Invalid choice! Please select 1 to 5.")
```
