```python
# Shopping Discount Calculator

def calculate_bill(amount, discount_type="normal"):

    if amount >= 10000:
        discount_percent = 20
    elif amount >= 5000:
        discount_percent = 10
    elif amount >= 2000:
        discount_percent = 5
    else:
        discount_percent = 0

    discount = amount * discount_percent / 100
    final_price = amount - discount

    print("Original Price:", amount)
    print("Discount:", discount)
    print("Final Price:", final_price)

calculate_bill(12000)
```
