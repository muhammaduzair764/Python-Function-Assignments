```python
# Ride Fare Calculator

def calculate_fare(distance, peak_hours):

    if distance <= 5:
        fare = 100

    elif distance <= 10:
        fare = 100 + (distance - 5) * 30

    else:
        fare = 100 + (5 * 30) + (distance - 10) * 25

    if peak_hours == "yes":
        fare = fare * 1.20

    print("Distance:", distance, "km")
    print("Peak Hours:", peak_hours)
    print("Total Fare: Rs.", fare)
distance = float(input("Enter Distance (km): "))
peak_hours = input("Is it peak hours? (yes/no): ")

calculate_fare(distance, peak_hours)
```
