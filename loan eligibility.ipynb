```python
def check_loan_eligibility(age, salary, credit_score):
    if age < 21:
        print("Loan Rejected")
        print("Reason: Age is below 21")
    elif salary < 50000:
        print("Loan Rejected")
        print("Reason: Salary is below 50,000")
    elif credit_score < 650:
        print("Loan Rejected")
        print("Reason: Credit score is below 650")
    else:
        print("Loan Approved")


age = int(input("Enter age: "))
salary = float(input("Enter monthly salary: "))
credit_score = int(input("Enter credit score: "))

check_loan_eligibility(age, salary, credit_score)
```
