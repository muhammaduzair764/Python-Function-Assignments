```python
# Mobile Package Billing:
def calculate_mobile_bill(package, extra_gb):

    # Package price
    if package == "Basic":
        price = 1000
        free_gb = 0
    elif package == "Standard":
        price = 1500
        free_gb = 2
    elif package == "Premium":
        price = 2500
        free_gb = 5
    else:
        return "Invalid package"

    # Extra data charge
    chargeable_gb = max(0, extra_gb - free_gb)
    extra_cost = chargeable_gb * 100

    # Final bill
    final_bill = price + extra_cost

    print("Package Price:", price)
    print("Free GB:", free_gb)
    print("Chargeable GB:", chargeable_gb)
    print("Extra Data Cost:", extra_cost)
    print("Final Bill:", final_bill)


# User input
package = input("Enter package (Basic/Standard/Premium): ")
extra_gb = float(input("Enter extra GB: "))

calculate_mobile_bill(package, extra_gb)
```
